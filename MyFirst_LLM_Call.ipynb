### First LLM Call with Langchain

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model = "openai/gpt-oss-120b",
    temperature=0
)


response = llm.invoke("What is  Langchain in one sentence")
print("AI Response: ", response.content)
```
