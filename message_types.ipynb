```python
from dotenv import load_dotenv
import os 
load_dotenv()
```

```python
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage, SystemMessage


def main():
    print("Understanding message types\n")

    
    model = ChatGroq(
        model="openai/gpt-oss-120b",
        temperature=0
    )
    # Using structured messages for better control
    messages = [
        SystemMessage(content="Your name is Alexa, and you are a senior LangChain tutor. Always introduce yourself by name and profession."),
        HumanMessage(content="what is LangChain in one sentence?")
    ]

    
    response = model.invoke(messages)

    print("🤖 AI Response:\n")
    print(response.content)
    print("\n✅ Notice how the SystemMessage influenced the response style!")



if __name__ == "__main__":
    main()


```

```python
system_messages = [
    "Your name is lora",
    "You are a helpful assistant",
    "You are a senior AI developer tutor",
    "You are s senior data analyst"
]

model1 = ChatGroq(
        model="openai/gpt-oss-120b",
        temperature=0
    )


messages = [
    SystemMessage(" ".join(system_messages)),
    HumanMessage("Tell me about yourself. How can you help me?")
]

response_1 = model1.invoke(messages)
print("AI Message: ", response_1.content)
```
