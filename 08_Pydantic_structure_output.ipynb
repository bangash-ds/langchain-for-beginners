### Structured Outputs with Pydantic

Use Pydantic models to get type-safe, structured data from LLMs. This ensures you get exactly the data structure you need.

The Schema Analogy:

Structured Output

Think of structured outputs like a form:

You define the fields (name, email, age)
The AI fills in the form
You get validated, typed data back

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq 
from pydantic import BaseModel, Field


# define the output schema with pydantic
class person(BaseModel):
    """schema for person information"""
    name: str = Field(description="The person's full name") 
    age: int = Field(description="The person's age in year")
    occupation: str = Field(description="The person's job or profession")

def main():
    llm = ChatGroq(
        model="openai/gpt-oss-120b",
        temperature=0

    )    
    # Create a structure output model
    structured_model = llm.with_structured_output(person)


    text1 = "John Smith is a 35-year-old software engineer from Seattle."
    text2 = "Abdul Salam famous in company as Javed 18 years old works in telecom"

    structured_response  = structured_model.invoke(f"Extract person's information from: {text1}")
    print(f"Name: {structured_response.name}")
    print(f"Age: {structured_response.age}")
    print(f"Occupation: {structured_response.occupation}")


if __name__ == "__main__":
    main()
```

### Example 8: Complex Pydantic Schemas
Build more sophisticated schemas with `nested objects`, `enums`, and `validation`.

```python
from langchain_groq import ChatGroq
from pydantic import Field, BaseModel
from typing import Literal


class Address(BaseModel):
    street: str = Field(description="street address")
    city: str = Field(description="city name")
    country: str = Field(description="country name")


class company(BaseModel):
    name: str = Field(description="company name")
    industry: Literal["Technology", "Finance", "Healthcare", "Retail", "other"] = Field(
        description="Industry sector"
        )

    employees_count: int = Field(description="Number of employees")
    headquarters: Address = Field(description="Companies headquarters location")
    is_public: bool = Field(description="Whether the company is publicly traded")


def main():
    llm1 = ChatGroq(
        model="openai/gpt-oss-120b",
        temperature=0

    )

    # Create structured model
    with_structure_model = llm1.with_structured_output(company)


    text3 = """
    Microsoft Corporation is a technology giant headquartered in Redmond, Washington, USA.
    The company has approximately 220,000 employees worldwide and is publicly traded on NASDAQ.
    """       

    result_model = with_structure_model.invoke(f"Extract the information from: {text3}")


    print(f"Campany name {result_model.name}")
    print(f"Industry {result_model.industry}")
    print(f"Employees {result_model.employees_count}")
    print(f"Location {result_model.headquarters.city}, {result_model.headquarters.country}")
    print(f"is_public {result_model.is_public}")


if __name__ == "__main__":
    main()    
```
