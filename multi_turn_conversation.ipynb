```python
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage


llm = ChatGroq(
    model = "openai/gpt-oss-120b",
    temperature=0
)

messages = [
        {'role': 'system', 'content':'Your name is Siri. Always introduce yourselfe by name and profession'},
        {'role': 'user', 'content':'Hi, Who are you and how can you help me?'}
]

response = llm.invoke(messages)

print(response.content)
print("-" * 50)
print(response.content_blocks)
print("-" * 50)
print([response.usage_metadata])
```

```python
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage

def main():
    print(" Multi-Turn Conversation Example\n")


    llm = ChatGroq(
        model = "openai/gpt-oss-120b",
        temperature=0
    )

    messages1 = [
        SystemMessage(content="You are a helpful coding tutor and Keep every answer under 100 words."),
        HumanMessage(content="What is Python?")
    ]

    print("User: what is Python:")

    # First exchange
    response1 = llm.invoke(messages1)
    print("\nAI: ",response1.content)
    messages1.append(AIMessage(content=response1.content))

    # Second exchange - AI remembers the context
    print("\ncan you show me a simple example")
    messages1.append(HumanMessage(content="can you show me a simple example"))


    response2 = llm.invoke(messages1)
    messages1.append(AIMessage(content=response2.content))
    print("\nAI: ", response2.content)


    # Third exchange - AI still remembers everything
    print("\nwhat are the benefits compared to other languages")
    messages1.append(HumanMessage(content="What are the benefits compare to other languages"))

    # the 3rd AI response is not added to conversation history since it is the last in the conversation
    response3 = llm.invoke(messages1)
    print("\nAI: ", response3.content)

    print("\n\n✅ Notice how the AI maintains context throughout the conversation!")
    print(f"📊 Total messages in history: {len(messages1)} messages, that include 1 system message, 3 Human messages and 2 AI responses")

if __name__ == "__main__":
    main()
```
