```python
from langchain_groq import ChatGroq


def temperature_comparison():

    prompt = "Write a creative opening line for a sci-fi story about time travel."
    temperatures = [0, 1, 2] # 0 -> deterministic, 1 -> Balanced, 2 -> Creative 


    for temp in temperatures:
        print("-" * 80)
        print(f"temp: {temp}")
       

        llm = ChatGroq(
            model="openai/gpt-oss-120b",
            temperature=temp
        )



        try:    
            for i in range(1, 3):
                response = llm.invoke(prompt)
                print(f"Try {i}: {response.content}")

        except Exception as e:
            # Some models may not support certain temperature values
            print(f"This model does not support the temperature={temp} Skipping...")
            print(f"Error: {e}")


    
print("\n General Temperature Guidelines:")
print("   - Lower values (0-0.3): More deterministic, consistent responses")
print("   - Medium values (0.7-1.0): Balanced creativity and consistency")
print("   - Higher values (1.5-2.0): More creative and varied responses")


if __name__ == "__main__":
    temperature_comparison()    
```

```python

def max_tokens_comparisons():

    prompt1 = "Explain how the internet works."
    token_limits = [100, 200, 300]


    for max_token in token_limits:
        print("-" * 80)
        print(f"max_tokens: {max_token}")

        model = ChatGroq(
            model="openai/gpt-oss-120b",
            temperature=0,
            max_tokens=max_token
        )


        response1 = model.invoke(prompt1)

        print(f"Characters: {len(response1.content)}")
        print(f"Tokens used: {response1.usage_metadata['output_tokens']}")
        print(f"Finish reason: {response1.response_metadata.get('finish_reason')}")
        print(f"Response: {response1.content}\n")


if __name__ == "__main__":
    max_tokens_comparisons()
```
